# Проект: Обучение с учителем: качество модели

###### Описание проекта

**Заказчик:** Интернет-магазин «В один клик» продаёт разные товары: для детей, для дома, мелкую бытовую технику, косметику и даже продукты. Отчёт магазина за прошлый период показал, что активность покупателей начала снижаться. Привлекать новых клиентов уже не так эффективно: о магазине и так знает большая часть целевой аудитории. Возможный выход — удерживать активность постоянных клиентов. Сделать это можно с помощью персонализированных предложений.

**Цель:** разработать решение, которое позволит персонализировать предложения постоянным клиентам, чтобы увеличить их покупательскую активность.

**Задачи:** 
- Нужно построить модель, которая предскажет вероятность снижения покупательской активности клиента в следующие три месяца.
- В исследование нужно включить дополнительные данные финансового департамента о прибыльности клиента: какой доход каждый покупатель приносил компании за последние три месяца.
- Используя данные модели и данные о прибыльности клиентов, нужно выделить сегменты покупателей и разработать для них персонализированные предложения.

**Данные:**

`market_file.csv`

Таблица, которая содержит данные о поведении покупателя на сайте, о коммуникациях с покупателем и его продуктовом поведении.
- `id` — номер покупателя в корпоративной базе данных.
- `Покупательская активность` — рассчитанный класс покупательской активности (целевой признак): «снизилась» или «прежний уровень».
- `Тип сервиса` — уровень сервиса, например «премиум» и «стандарт».
- `Разрешить сообщать` — информация о том, можно ли присылать покупателю дополнительные предложения о товаре. Согласие на это даёт покупатель.
- `Маркет_актив_6_мес` — среднемесячное значение маркетинговых коммуникаций компании, которое приходилось на покупателя за последние 6 месяцев. Это значение показывает, какое число рассылок, звонков, показов рекламы и прочего приходилось на клиента.
- `Маркет_актив_тек_мес` — количество маркетинговых коммуникаций в текущем месяце.
- `Длительность` — значение, которое показывает, сколько дней прошло с момента регистрации покупателя на сайте.
- `Акционные_покупки` — среднемесячная доля покупок по акции от общего числа покупок за последние 6 месяцев.
- `Популярная_категория` — самая популярная категория товаров у покупателя за последние 6 месяцев.
- `Средний_просмотр_категорий_за_визит` — показывает, сколько в среднем категорий покупатель просмотрел за визит в течение последнего месяца.
- `Неоплаченные_продукты_штук_квартал` — общее число неоплаченных товаров в корзине за последние 3 месяца.
- `Ошибка_сервиса` — число сбоев, которые коснулись покупателя во время посещения сайта.
- `Страниц_за_визит` — среднее количество страниц, которые просмотрел покупатель за один визит на сайт за последние 3 месяца.

`market_money.csv`

Таблица с данными о выручке, которую получает магазин с покупателя, то есть сколько покупатель всего потратил за период взаимодействия с сайтом.
- `id` — номер покупателя в корпоративной базе данных.
- `Период` — название периода, во время которого зафиксирована выручка. Например, 'текущий_месяц' или 'предыдущий_месяц'.
- `Выручка` — сумма выручки за период.

`market_time.csv`

Таблица с данными о времени (в минутах), которое покупатель провёл на сайте в течение периода.
- `id` — номер покупателя в корпоративной базе данных.
- `Период` — название периода, во время которого зафиксировано общее время.
- `минут` — значение времени, проведённого на сайте, в минутах.

`money.csv`

Таблица с данными о среднемесячной прибыли покупателя за последние 3 месяца: какую прибыль получает магазин от продаж каждому покупателю.
- `id` — номер покупателя в корпоративной базе данных.
- `Прибыль` — значение прибыли.

## Исследовательский анализ

Создадим функцию для построения боксплота и гистограммы для нужного числового признака. Используем библиотеку `seaborn`.

```python
def show_box_hist_plots (
    col,
    bins = 100,
    title = 'Распределение значений',
    xlabel = None
):
    
    fig, ax = plt.subplots(2, sharex = True, gridspec_kw={"height_ratios": (.15, .85)})
    sns.histplot(x = col, ax = ax[1], bins = bins, color = '#ff7f0e')
    sns.boxplot(x = col, ax = ax[0], orient = 'h', color = '#ff7f0e')
    ax[0].set(title = title)
    ax[1].set(ylabel = 'Количество значений, шт', xlabel = xlabel)
    plt.subplots_adjust(top=0.8)
    plt.show()

    print(f'Среднее: {col.mean()}')
    print(f'Медиана: {col.median()}')
    print(f'Мода: {col.mode()[0]}')
    print(f'Стандартное отклонение: {col.std()}')
```

Аналогично числовым признакам создадим функцию для категориальных, она будет строить столбчатую и круговую диаграммы.

```python
def show_barplot_pie (col, title = 'Диаграмма'):
    # создадим фигуру
    fig, ax = plt.subplots(1, 2, figsize=(8,4))
    
    # передадим в переменную распределение значений
    value_counts_result = col.value_counts().sort_index() 
    
    # определим цвет колонок
    colors = sns.color_palette("Paired", len(value_counts_result))
    
    # создадим столбчатой диаграммы
    sns.barplot(
        x=value_counts_result.index,
        hue=value_counts_result.index,
        y=value_counts_result.values,
        ax = ax[0],
        palette = colors
    )
    # зададим подпись оси Y
    ax[0].set(ylabel = 'Количество значений, шт', xlabel = None)
    
    # отображение значений для каждого столбца
    for index, value in enumerate(value_counts_result):
        ax[0].text(index, 40, str(value), ha='center', fontsize=16, color='#000000')
        
    # создание круговой диаграммы
    ax[1].pie(
        value_counts_result,
        labels = value_counts_result.index,
        autopct='%1.1f%%',
        startangle=90,
        colors=colors
    )
    # подпись каждой диаграммы
    ax[0].set_title('Столбчатая диаграмма', fontsize=12)
    ax[1].set_title('Круговая диаграмма', fontsize=12)
    fig.suptitle(title, fontsize=16)
    plt.subplots_adjust(top=0.8)
    plt.tight_layout()
    
    # уберем легенду, если она существует
    legend = ax[0].get_legend()
    if legend:
        legend.remove()
    
    plt.show()
```

## Корреляционный анализ

Для проведения корреляционного анализа используем библиотеку phik. Преимуществом этого подхода является возможность проверять взаимосвязь признаков с разными типами данных.

Целевой признак - покупательская активность

```python
def create_phik_matrix(df, interval_cols):
    phik_overview = phik_matrix(df, interval_cols=interval_cols) 
    plot_correlation_matrix(
        phik_overview.values,
        x_labels=phik_overview.columns,
        y_labels=phik_overview.index,
        vmin=0, vmax=1, color_map='Blues',
        title=r'correlation $\phi_K$',
        fontsize_factor=1.5,
        figsize=(20, 15)
    )
```

```python
create_phik_matrix(total, interval_cols = total.select_dtypes(include=['int', 'float']).columns)
```

- Полезный признак `акционные покупки` коррелирует с `выручкой за предыдущий и препредыдущий месяца`. Причем между собой эти две выручки тоже связаны. Избавимся от этих двух признаков. 
- Как показал исследовательский анализ признаки `минут_пред` и `минут_тек` имеют одинаковое распределение. Ввиду меньшей корреляцией с другими признаками оставим `минут_тек`.
-`id` не включаем в матрицу.

Пар признаков, чей коэффицент корреляции больше 0.9, не обнаружено. Поэтому можно сказать, что мультиколлинеарности нет. Однако есть ряд признаков с околонулевой связью с целевым признаком:
- `длительность`, 
- `маркет_актив_тек_мес`, 
- `разрешить_сообщать`,
- `ошибка_сервиса` - эмпирическим путем выяснено, что данный признак мешает модели.

Уберем эти признаки для снижения шума.

## Выбор модели

### Подготовка данных

Создадим переменные с названиями признаков для соответствующей обработки.

```python
num_cols = [
    'маркет_актив_6_мес',
    'акционные_покупки',
    'средний_просмотр_категорий_за_визит',
    'неоплаченные_продукты_штук_квартал',
    'страниц_за_визит',
    'минут_тек',
    'выручка_тек',
    'выручка_пред', 
    'выручка_препред', 
    'минут_пред' 
]

ohe_cols = ['популярная_категория'] 
ord_cols = ['тип_сервиса']
```

---

Разделим получившийся датасет `total` на выборки.

```python
X = total.drop(['покупательская_активность'], axis=1)
y = total['покупательская_активность']

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.25,
    random_state=RANDOM_STATE,
    stratify=y
)
[i.shape for i in [X_train, X_test, y_train, y_test]]
```

Целевой признак закодируем с помощью `LabelEncoder`. Проверим размер тестовых и тренировочных выборок.

```python
le = LabelEncoder()

y_train = le.fit_transform(y_train)
y_test = le.transform(y_test)

print(f'Порядок кодирования целевого признака: {le.classes_}.') 
```

Приступим к созданию пайплана. Используем `ColumnTransformer` для применения различных методов обработки к разным столбцам датафрейма в рамках одного пайплайна.

```python
preprocessor = ColumnTransformer(
    [
        ('ord', OrdinalEncoder(categories=[['стандарт', 'премиум']]), ord_cols),
        ('ohe', OneHotEncoder(sparse_output=False, drop='first', handle_unknown='ignore'), ohe_cols),
        ('scaler', StandardScaler(), num_cols)
    ],
    remainder='passthrough'
)
```

Пайплайн для подготовки данных готов. Объединим его с блоком выбора модели.

```python
pipe_final = Pipeline(
    [
        ('preprocessor', preprocessor),
        ('models', DecisionTreeClassifier(random_state=RANDOM_STATE))
    ]
)
```

### Лучшая модель

Для решения задачи классификации переберем 4 модели с собственными гиперпараметрами:
- KNeighborsClassifier
- DecisionTreeClassifier
- SVC
- LogisticRegression
 
Количественные признаки будут масштабироваться двумя способами:
- StandardScaler
- MinMaxScaler

Ввиду дисбаланса целевого признака, показанного в исследовательском анализе, добавим в модели SVC и Lin.Reg.  параметр `class_weight = 'balanced'`.

```python
param_distributions = [
    # словарь для модели KNeighborsClassifier() 
    {
        'models': [KNeighborsClassifier()],
        'models__n_neighbors': range(1, 20), # количество ближайших соседей
        'models__metric': ['euclidean', 'manhattan'], # метрика расстояния
        'preprocessor__scaler': [StandardScaler(), MinMaxScaler()]   # масштабирование num признаков
    },
    # словарь для модели DecisionTreeClassifier()
    {
        'models': [DecisionTreeClassifier(random_state=RANDOM_STATE)],
        'models__max_depth': range(2, 11), # глубина дерева
        'models__min_samples_leaf': range(1, 11), # минимум образцов для разбиения
        'models__min_samples_split': range(2, 11), # минимум образцов в листе
        'preprocessor__scaler': [StandardScaler(), MinMaxScaler()]  # масштабирование num признаков
    },
    # словарь для модели SVC()
    {
        'models': [SVC(class_weight='balanced', random_state=RANDOM_STATE, probability=True)], 
        'models__C': [0.01, 0.1, 1, 10, 100], # сила регуляризации
        'models__kernel': ['linear', 'rbf', 'sigmoid'], # тип ядра
        'preprocessor__scaler': [StandardScaler(), MinMaxScaler()]  # масштабирование num признаков

    },
        {
        'models': [SVC(class_weight='balanced', kernel='poly', random_state=RANDOM_STATE, probability=True)], 
        'models__C': [0.01, 0.1, 1, 10, 100], # сила регуляризации
        'models__degree': [1, 2, 3, 4, 5], # степень полиномиального ядра
        'preprocessor__scaler': [StandardScaler(), MinMaxScaler()]  # масштабирование num признаков

    },
    # словари для модели LogisticRegression()
    {
        'models': [LogisticRegression(class_weight='balanced', max_iter=500, solver='liblinear', random_state=RANDOM_STATE)],
        'models__C': [0.01, 0.1, 1, 10, 100], # сила регуляризации
        'models__penalty': ['l1', 'l2'], # тип регуляризации
        'preprocessor__scaler': [StandardScaler(), MinMaxScaler()] # масштабирование num признаков
    },
    {
        'models': [LogisticRegression(class_weight='balanced', max_iter=500, solver='lbfgs', random_state=RANDOM_STATE)],
        'models__C': [0.01, 0.1, 1, 10, 100], # сила регуляризации
        'models__penalty': ['l2'], # тип регуляризации
        'preprocessor__scaler': [StandardScaler(), MinMaxScaler()] # масштабирование num признаков
    },
    {
        'models': [LogisticRegression(class_weight='balanced', max_iter=500, solver='saga', random_state=RANDOM_STATE)],
        'models__C': [0.01, 0.1, 1, 10, 100], # сила регуляризации
        'models__penalty': ['l1', 'l2'], # тип регуляризации
        'preprocessor__scaler': [StandardScaler(), MinMaxScaler()] # масштабирование num признаков
    }
] 
```

Для перебора гиперпараметров и моделей используем метод `GridSearch`, который пройдется по всем возможным комбинациям. 

Выбор между `F1` и `ROC-AUC`:
Ввиду того, что нас интересует общая предсказательная способность модели, а не конкретный баланс между `precision` и `recall` выбираем `ROC-AUC`.  Причем присутвует умеренный дисбаланс классов, к которому выбранная метрика нечувствительна.

```python
grid = GridSearchCV(
    pipe_final, 
    param_grid=param_distributions, 
    cv=5, 
    scoring='roc_auc',
    n_jobs=-1,
    error_score='raise'
)
```

```python
grid.fit(X_train, y_train)
```

Лучшая модель и её параметры:

```python
grid.best_params_
```

```python
print('Метрика ROC-AUC на тренировочной выборке:', grid.best_score_)
y_pred = grid.best_estimator_.predict_proba(X_test)[:, 1]
print(f'Метрика ROC-AUC на тестовой выборке: {roc_auc_score(y_test, y_pred)}')
print(f'Разброс между данными {abs(grid.best_score_ - roc_auc_score(y_test, y_pred))}')
```

**Вывод:**
1) Наилучшая модель - `SVC`;

2) Гиперпараметры: 
  - Сила регуляризации: `С=1`,
  - Ядро: `rbf`,
  - Обработка дисбаланса:` class_weight='balanced'`
 
3) Количественные признаки `стандартизированны`.

```python
sns.scatterplot(
    segmentation_df,
    y = 'минут_тек',
    x = 'страниц_за_визит',
    hue = 'вероятность_снижения_new'
)
plt.axhline(y=13.5, c='r')
plt.axvline(x=7.5, c='r')
plt.title('Диаграмма рессеяния проведенного на сайте времени \nв текущем месяце от кол-ва посещенных страниц за визит.')
plt.show()
```

Сниженная активность на сайте:
- Пользователь посетил менее 8 страниц нас сайте;
- Пользователь провел не более 13 минут на сайте.

```python
site_activity = ( 
    segmentation_df
    .query('минут_тек < 13.5 & страниц_за_визит < 8')['вероятность_снижения_new']
    .sum() 
    / segmentation_df
    .query('минут_тек < 13.5 & страниц_за_визит < 8')['вероятность_снижения_new']
    .count()
).round(3)*100

print(f'Доля клиентов со сниженными покупательской активностью и низкой активности на сайте: {site_activity}%')

site_activity_share = (
    segmentation_df
    .query("минут_тек < 13.5 & страниц_за_визит < 8")['тип_сервиса']
    .count() 
    / 
    segmentation_df["минут_тек"].count()
).round(3)*100

print(f'Процент пользователей c низкой активностью на сайте: {site_activity_share}%.')
```

Данный сегмент оказался самым обширным - 32% от всего числа пользователей, причем 84% из них - пользователи со сниженной покупательской активностью.

# Проект: HR-аналитика

<span style="font-size:20px; font-weight:500;">Описание проекта:</span>

**Заказчик:** Компания «Работа с заботой», HR-аналитики которой помогают бизнесу оптимизировать управление персоналом: бизнес предоставляет данные, а аналитики предлагают, как избежать финансовых потерь и оттока сотрудников.

**Цель:** разработать решение с применением машинного обучения, с помощью которого получится быстрее и точнее отвечать на вопросы бизнеса.

**Задачи:** 
1. Построить модель, которая сможет предсказать уровень удовлетворённости сотрудника.
2. Построить модель, которая сможет предсказать то, что сотрудник уволится из компании.

**Данные:** Компания предоставила данные с характеристиками сотрудников компании. Среди них — уровень удовлетворённости сотрудника работой в компании. Эту информацию получили из форм обратной связи: сотрудники заполняют тест-опросник, и по его результатам рассчитывается доля их удовлетворённости от 0 до 1, где 0 — совершенно неудовлетворён, 1 — полностью удовлетворён.

- `id` — уникальный идентификатор сотрудника;
- `dept` — отдел, в котором работает сотрудник;
- `level` — уровень занимаемой должности;
- `workload` — уровень загруженности сотрудника;
- `employment_years` — длительность работы в компании (в годах);
- `last_year_promo` — показывает, было ли повышение за последний год;
- `last_year_violations` — показывает, нарушал ли сотрудник трудовой договор за последний год;
- `supervisor_evaluation` — оценка качества работы сотрудника, которую дал руководитель;
- `salary` — ежемесячная зарплата сотрудника;
- `job_satisfaction_rate` — уровень удовлетворённости сотрудника работой в компании, целевой признак;
- `quit` - показывает, уволился ли сотрудник.

**Датасеты для задачи №1:**

### Предобработка

Создадим функцию, которая выведет основную информацию о датафрейме:
- Размер
- Кол-во строк дубликатов
- Тип значений в столбцах
- Кол-во ненулевых значений
- Кол-во уникальных значений
- Список кникальных значений

```python
def df_info(df):
    df_ = pd.DataFrame(
        index = df.columns,
        data = {
            "dtype": [df[col].dtype for col in df.columns],
            "non_null": [df[col].notna().sum() for col in df.columns],
            "unique_count": [df[col].nunique() for col in df.columns],
            "unique_values": [df[col].unique() for col in df.columns]
        })
    print(f' Размер датафрейма: {df.shape}')
    print(f' Кол-во строк-дубликатов: {df.duplicated().sum()}')
    return df_
```

Создадим функцию, которая проверит вхождение значений признака id в оба переданных датафрейма. Необходимо, чтобы в обучающей и тестовой выборках номера сотрудников не повторялись.

```python
def value_comparison(df_1, df_2, col):
    result = df_1[col].isin(list(df_2[col])).sum()
    if result == 0:
        print(f'Схожих значений по признаку {col} не найдено. Выборки не пересекаются.')
    else:
        print(f'По признаку {col} найдено {result} схожих значений. Выборки пересекаются.')
```

```python
value_comparison(test_features, train_job_satisfaction, 'id')
```

```python
def show_plots (
    df_1,
    df_2,
    category = Literal['num', 'cat'],
    col=None,
    bins = 100,
    title = 'Распределение значений',
    xlabel = None
):
    if category == 'num':
        
        # создадим фигуру 2х2
        fig, ax = plt.subplots(2, 2, sharex = True, gridspec_kw={"height_ratios": (.15, .85)}, figsize=(10,6))
        
        for ax_, df, name in zip([ax[0,0], ax[0,1]], [df_1, df_2], ['Тренировочная выборка', 'Тестовая выборка']):
            sns.boxplot(df[col], ax = ax_, orient = 'h', color = '#ff7f0e')
            ax_.set_title(name, fontsize=16)
            
        for ax_, df in zip([ax[1,0], ax[1,1]], [df_1, df_2]):   
            sns.histplot(df[col], ax = ax_, bins = bins, color = '#ff7f0e')
            ax_.set(ylabel = 'Количество значений, шт', xlabel = xlabel)
            
        fig.suptitle(title, fontsize=18)
        plt.subplots_adjust(top=0.85, wspace=0.3)
        return plt.show()

        #print(f'Среднее: {col.mean()}')
        #print(f'Медиана: {col.median()}')
        #print(f'Мода: {col.mode()[0]}')
        #print(f'Стандартное отклонение: {col.std()}')
        
    elif category == 'cat':

        # создадим фигуру 2х2
        fig, ax = plt.subplots(2, 2, figsize=(10, 9))

        for ax_, df, name in zip([ax[0,0], ax[0,1]], [df_1, df_2], ['Тренировочная выборка', 'Тестовая выборка']):

            # передадим в переменную распределение значений
            value_counts = df[col].value_counts().sort_index()

            # создадим столбчатой диаграммы
            sns.barplot(
                    x=value_counts.index,
                    hue=value_counts.index,
                    y=value_counts.values,
                    ax = ax_,
                    palette = sns.color_palette('coolwarm', len(value_counts))
                )    

            # зададим подпись оси Y
            ax_.set(ylabel = 'Количество значений, шт', xlabel = None)  

            # повернем названия категорий столбчатой диаграммы
            plt.setp(ax_.get_xticklabels(), rotation=90)    

            # отображение значений для каждого столбца
            for index, value in enumerate(value_counts):
                ax_.text(index, max(value_counts) * 0.05, str(value), ha='center', color='#000000')

            # подпись каждой диаграммы
            ax_.set_title(name, fontsize=16)
            #ax_.set(title = name, fontsize=16)
            
            # уберем легенду, если она существует
            legend = ax_.get_legend()
            if legend:
                legend.remove()

        for ax_, df in zip([ax[1,0], ax[1,1]], [df_1, df_2]):

            # передадим в переменную распределение значений
            value_counts = df[col].value_counts().sort_index()

            # создание круговой диаграммы
            ax_.pie(
                value_counts,
                labels = value_counts.index,
                autopct='%1.1f%%',
                startangle=90,
                colors=sns.color_palette('coolwarm', len(value_counts))
            )

            # подпись каждой диаграммы
            #ax_.set_title('Круговая диаграмма', fontsize=12)
            
            # уберем легенду, если она существует
            legend = ax_.get_legend()
            if legend:
                legend.remove()

        fig.suptitle(title, fontsize=18)
        plt.subplots_adjust(top=0.9, wspace=0.3, hspace=0.3)
        
        return plt.show()
```

```python
def box_upper_bound(df, col):  
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    upper_bound = Q3 + IQR * 1.5
    return upper_bound
```

Создадим переменные с названиями признаков для соответствующей обработки. Признаки `employment_years` и `supervisor_evaluation` оставим без изменения.

```python
std_cols = ['salary']
ohe_cols = ['dept', 'last_year_promo', 'last_year_violations']
ord_cols = ['level', 'workload']
```

Ввиду того, что в тренировочной и тестовой выборках все категориальные признаки имеют одинаковые категории, необходимости добалять второй SimpleImputer нет. Пропуски, которые сейчас присутствуют в признаках заменятся первым импутером на моду.

```python
ord_pipe = Pipeline(
    [
        (
            'imputer_before_ord', 
            SimpleImputer(missing_values=np.nan, strategy='most_frequent')
        ),
        (
            'ord',  
            OrdinalEncoder(categories=[['junior', 'middle', 'senior'], ['low', 'medium', 'high']])
        )
    ]
)
```

Аналогично для категориальных номинальных используем комбинацию SimpleImputer + энкодер.

```python
ohe_pipe = Pipeline(
    [
        (
            'imputer_before_ohe',
            SimpleImputer(missing_values=np.nan, strategy='most_frequent')
        ), 
        (
            'ohe',
            OneHotEncoder(drop='first', sparse_output=False)
        )
    
    ]
)
```

Используем `ColumnTransformer` для применения различных методов обработки к разным столбцам датафрейма в рамках одного пайплайна.

```python
preprocessor = ColumnTransformer(
    [
        ('ord', ord_pipe, ord_cols),
        ('ohe', ohe_pipe, ohe_cols),
        ('std', StandardScaler(), std_cols)
    ],
    remainder='passthrough'
)
```

Пайплайн для подготовки данных готов. Объединим его с блоком выбора модели.

```python
pipe_final = Pipeline(
    [
        ('preprocessor', preprocessor),
        ('models', DecisionTreeRegressor(random_state=RANDOM_STATE))
    ]
)
```

```python
def smape(y_test, y_pred):
    return 100 * np.mean(np.abs(y_test - y_pred) / ((np.abs(y_test) + np.abs(y_pred)) / 2))

smape_score = make_scorer(smape, greater_is_better=False)
```

Для решения задачи регрессии переберем 3 модели с собсвенными гиперпараметрами:
- DecisionTreeRegressor (дерево решения)
- Ridge (линейная регрессия с L2 регуляризацией)
- KernelRidge (ядерная регрессия)

```python
params = [
    # словарь для модели DecisionTreeRegressor() 
    {
        'models': [DecisionTreeRegressor(random_state=RANDOM_STATE)],
        'models__max_depth': range(1,11),  # Ограничение глубины дерева
        'models__min_samples_split': range(2,10),  # Минимальное число объектов для разбиения узла
        'models__min_samples_leaf': range(1,6),  # Минимальное число объектов в листе
    },
    # словарь для модели Ridge() 
    {
        'models': [Ridge(random_state=RANDOM_STATE)],
        'models__alpha': [0.01, 0.1, 1, 10, 100],  # Коэффициент регуляризации
    },
    # словарь для модели KernelRidge() 
    {
        'models': [KernelRidge()],
        'models__alpha': [0.01, 0.1, 1, 10],  # Коэффициент регуляризации
        'models__kernel': ['linear', 'rbf'],  # Только линейное и гауссовское ядро
        'models__gamma': [0.1, 1, 3],  # Параметр ядра (используется для RBF)
    }
]
```

Для перебора гиперпараметров и моделей используем метод `GridSearch`, который пройдется по всем возможным комбинациям. 

```python
grid = GridSearchCV(
    pipe_final, 
    param_grid=params, 
    cv=5, 
    scoring=smape_score,
    n_jobs=-1,
    error_score='raise'
)
```

#### Топ-5

```python
def top_models(grid):
    top = (
        pd.DataFrame(grid.cv_results_)
        .sort_values(by = 'rank_test_score')
        .set_index('rank_test_score')
        [["param_models", "params", "mean_test_score", "std_test_score"]]
        .head()
    )
    return top
```

```python
top_5_reg = top_models(grid)
top_5_reg['mean_test_score'] = top_5_reg['mean_test_score'] * -1
top_5_reg
```

Первые две модели - ядерная регрессия, далее - деревья решений. Модели ядерной регрессии отличаются только коэффициентом регуляризации. Деревья отличаются только минимальным кол-вом листьев для разбиения. Посмотрим, не произошло ли переобучения у деревьев решения. Для этого получим предсказания на тестовой выборке каждой модели со своими гиперпараметрами и посмотрим скор метрики smape. Для обработки признаков достанем из пайплайна обученный preprocessor.

```python
preprocessor_unit = grid.best_estimator_.named_steps['preprocessor']

X_train_transformed = preprocessor_unit.transform(X_train)
X_test_transformed = preprocessor_unit.transform(X_test)
```

```python
# Список для хранения SMAPE
smape_scores = []

# Обучаем и предсказываем для каждой модели
for row in top_5_reg.itertuples():
    params = row.params.copy()  # Берем параметры модели
    model_instance = params.pop("models")  # Берем объект модели
    model_class = type(model_instance)  # Получаем класс модели

    # Убираем 'models__' из ключей параметров
    params_cleaned = {k.replace("models__", ""): v for k, v in params.items()}
    
    # Если модель SVC, добавляем probability=True
    if model_class is not KernelRidge:
        params_cleaned["random_state"] = RANDOM_STATE

    # Создаем и обучаем модель
    model = model_class(**params_cleaned)
    model.fit(X_train_transformed, y_train)
    
    # Делаем предсказания
    y_pred = model.predict(X_test_transformed)
    
    # Вычисляем SMAPE и сохраняем
    smape_scores.append(smape(y_test, y_pred))
    
    
# Преобразуем результаты в pandas Series
smape_series = pd.Series(smape_scores, index=range(1, 6), name="SMAPE")

smape_series
```

```python
top_5_reg.join(smape_series)
```

Переобучения нет, все топ-5 моделей соответствуют требованиям заказчика: `SMAPE ≤ 15 на тестовой выборке`.

### Моделирование

#### Обучение модели

Для решения задачи классификации переберем 4 модели с собственными гиперпараметрами:
- KNeighborsClassifier
- DecisionTreeClassifier
- SVC
- LogisticRegression


Ввиду дисбаланса целевого признака, показанного в исследовательском анализе, добавим параметр `class_weight = 'balanced'`, там где это возможно.

```python
params_2 = [
    # словарь для модели KNeighborsClassifier() 
    {
        'models': [KNeighborsClassifier()],
        'models__n_neighbors': range(1, 15), # количество ближайших соседей
        'models__metric': ['euclidean', 'manhattan'], # метрика расстояния
    },
    # словарь для модели DecisionTreeClassifier()
    {
        'models': [DecisionTreeClassifier(random_state=RANDOM_STATE, class_weight='balanced')],
        'models__max_depth': [3, 5, 7, 10, 20, None], # глубина дерева
        'models__min_samples_leaf': [1, 2, 5, 10], # минимум образцов для разбиения
        'models__min_samples_split': [2, 5, 10, 20], # минимум образцов в листе
    },
    # словарь для модели SVC()
    {
        'models': [SVC(class_weight='balanced', random_state=RANDOM_STATE, probability=True)], 
        'models__C': [0.01, 0.1, 1, 10, 100], # сила регуляризации
        'models__kernel': ['linear', 'rbf', 'sigmoid'], # тип ядра
    },
        {
        'models': [SVC(class_weight='balanced', kernel='poly', random_state=RANDOM_STATE, probability=True)], 
        'models__C': [0.01, 0.1, 1, 10, 100], # сила регуляризации
        'models__degree': [1, 2, 3, 4, 5], # степень полиномиального ядра
    },
    # словари для модели LogisticRegression()
    {
        'models': [LogisticRegression(class_weight='balanced', solver = 'liblinear', random_state=RANDOM_STATE)],
        'models__C': [0.01, 0.1, 1, 10, 100], # сила регуляризации
        'models__penalty': ['l1', 'l2'] # тип регуляризации
    }
] 
```

Для перебора гиперпараметров и моделей используем метод `GridSearch`, который пройдется по всем возможным комбинациям. Будем пользоваться метрикой `ROC-AUC`.

```python
grid_2 = GridSearchCV(
    pipe_final_2, 
    param_grid=params_2, 
    cv=5, 
    scoring='roc_auc',
    n_jobs=-1,
    error_score='raise'
)
```

#### Топ-5

Применим функцию для сортировки и отображения 5 лучших моделей.

Подготовим данные для тестирования моделей на тестовой выборке.

```python
preprocessor_unit_2 = grid_2.best_estimator_.named_steps['preprocessor']

X_train_transformed_2 = preprocessor_unit_2.transform(X_train_2)
X_test_transformed_2 = preprocessor_unit_2.transform(X_test_2)
```

Добавим значения метрики на тестовых данных в таблицу.

```python
# Список для хранения ROC-AUC
roc_auc_scores = []

# Обучаем и предсказываем для каждой модели
for row in top_5_clf.itertuples():
    params = row.params.copy()  # Берем параметры модели
    model_instance = params.pop("models")  # Берем объект модели
    model_class = type(model_instance)  # Получаем класс модели

    # Убираем 'models__' из ключей параметров
    params_cleaned = {k.replace("models__", ""): v for k, v in params.items()}
    
    
    # Если модель SVC, добавляем probability=True
    if model_class is SVC:
        params_cleaned["probability"] = True
        
        # Проверяем, есть ли 'kernel=poly'
        if "kernel='poly'" in str(row.param_models):  
            params_cleaned['kernel'] = 'poly' 
            
    # Если модель не KNN, добавляем некоторые параметры
    if model_class is not KNeighborsClassifier:
        params_cleaned['class_weight'] = 'balanced'
        params_cleaned['random_state'] = RANDOM_STATE
        
    # Создаем и обучаем модель
    model = model_class(**params_cleaned)
    model.fit(X_train_transformed_2, y_train_2)
    
    # Делаем предсказания
    y_pred = model.predict_proba(X_test_transformed_2)[:,1]
    
    # Вычисляем ROC-AUC и сохраняем
    roc_auc_scores.append(roc_auc_score(y_test_2, y_pred))
    
    
# Преобразуем результаты в pandas Series
roc_auc_series = pd.Series(roc_auc_scores, index=range(1, 6), name="ROC-AUC")

# Добавим значения ROC-AUC в таблицу
top_5_clf = top_5_clf.join(roc_auc_series)

top_5_clf
```

Переобучения нет, все топ-5 моделей соответствуют требованиям заказчика: `ROC-AUC ≥ 0.91 на тестовой выборке`. Интересно, что вторая модель с коэффициентов регуляризации равным 10 на тестовых данных справилась немного лучше.
